# Count occurrences using find() in a loop

```python
text = "Hello World. Welcome to the World of Python. World is amazing."
sub = "World"
count = 0
index = text.find(sub)
while index != -1:
    print(f"Found '{sub}' at index {index}")
    count += 1
    index = text.find(sub, index + 1)
print(f"Total occurrences of '{sub}': {count}")
```

```python
text = "Hello World. Welcome to the World of Python. World is amazing."
sub = "World"
count = 0

for i in range(len(text)):
    index = text.find(sub, i)
    if index == i:
        print(f"Found '{sub}' at index {index}")
        count += 1

print(f"Total occurrences of '{sub}': {count}")
```
