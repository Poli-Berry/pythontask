```python
text = "banana banana banana"
needle = "na"

print(f"find({needle!r}):", text.find(needle))
print(f"rfind({needle!r}):", text.rfind(needle))
print(f"find({needle!r}, 3):", text.find(needle, 3))
print(f"rfind({needle!r}, 0, 10):", text.rfind(needle, 0, 10))
print(f"count({needle!r}):", text.count(needle))

positions = []
start = 0
while (pos := text.find(needle, start)) != -1:
    positions.append(pos)
    start = pos + 1

print(f"All positions of {needle!r}:", positions)
```
