```python
text = "banana banana banana"

print("find('na'):", text.find("na"))
print("rfind('na'):", text.rfind("na"))
print("find('na', 3):", text.find("na", 3))
print("rfind('na', 0, 10):", text.rfind("na", 0, 10))
print("count('na'):", text.count("na"))

positions = []
start = 0
while True:
    pos = text.find("na", start)
    if pos == -1:
        break
    positions.append(pos)
    start = pos + 1
print("All positions of 'na':", positions)
```
