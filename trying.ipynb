```python
# Integer
age = 25

# Float
price = 19.99

# String
name = "Alice"

# Boolean
is_student = True

# NoneType
nothing = None

# Complex number
z = 3 + 4j

# List (mutable sequence)
fruits = ["apple", "banana", "cherry"]

# Tuple (immutable sequence)
coordinates = (10.5, 20.3)

# Dictionary (key-value pairs)
person = {"name": "Bob", "age": 30}

# Set (unique unordered items)
unique_numbers = {1, 2, 3, 3, 2}

print("age:", age, type(age))
print("price:", price, type(price))
print("name:", name, type(name))
print("is_student:", is_student, type(is_student))
print("nothing:", nothing, type(nothing))
print("z:", z, type(z))
print("fruits:", fruits, type(fruits))
print("coordinates:", coordinates, type(coordinates))
print("person:", person, type(person))
print("unique_numbers:", unique_numbers, type(unique_numbers))
```

```python
name = input("Enter your name: ")
age = int(input("Enter your age: "))
height = float(input("Enter your height in meters: "))

birth_year = 2026 - age
bmi = 70 / (height ** 2)   # assuming 70 kg

print(f"Hello, {name}!")
print(f"You were born around {birth_year}.")
print(f"Your approximate BMI is {bmi:.2f}.")
print("Age as string:", str(age), type(str(age)))
print("Height as integer:", int(height), type(int(height)))
```

```python
a = 17
b = 5

print("Sum:", a + b)
print("Difference:", a - b)
print("Product:", a * b)
print("Division (float):", a / b)
print("Floor division:", a // b)
print("Modulus:", a % b)
print("Power:", a ** b)

print("a as float:", float(a))
print("b as string:", str(b))
print("'42' as int:", int("42"))
print("Round 3.14159 to 2 decimals:", round(3.14159, 2))
```

```python
first = "Hello"
second = "World"

greeting = first + ", " + second + "!"
print(greeting)
print(greeting.upper())
print(greeting.lower())
print("Length:", len(greeting))
print(f"Formatted: {first} — {second}")
print("Reversed:", greeting[::-1])
```

```python
my_list = [3, 1, 4, 1, 5, 9, 2, 6]
my_list.append(7)
my_list.sort()
print("List:", my_list)

my_tuple = (10, 20, 30)
print("Tuple:", my_tuple, "first element:", my_tuple[0])

my_set = {1, 2, 2, 3, 4, 4, 5}
print("Set (no duplicates):", my_set)

my_dict = {"apple": 3, "banana": 5, "cherry": 7}
for key, value in my_dict.items():
    print(f"{key}: {value}")
```

```python
x = 10
y = 20

print("x < y:", x < y)
print("x == y:", x == y)
print("x != y:", x != y)
print("x > 5 and y > 15:", x > 5 and y > 15)
print("x > 100 or y > 15:", x > 100 or y > 15)
print("not (x > y):", not (x > y))
```

```python
values = [42, 3.14, "100", True]

for v in values:
    print(f"Value: {v!r:8}  Type: {type(v).__name__}")

print("\nConversions:")
print(int("100") + 1)          # string → int
print(float("3.14") * 2)       # string → float
print(str(42) + "!")           # int → string
print(bool(0), bool(1), bool(""))   # various → bool
```
